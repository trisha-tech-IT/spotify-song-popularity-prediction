```python
# Calculate IQR for 'speechiness'
Q1 = df["speechiness"].quantile(0.25)
Q3 = df["speechiness"].quantile(0.75)
IQR = Q3 - Q1

# Define upper and lower limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# (Optional) Calculate original outliers to fix the second NameError on print(original_outliers)
original_outliers = df[(df["speechiness"] < lower_limit) | (df["speechiness"] > upper_limit)].shape[0]
```

**Model Evaluation**

**Model Evaluation Function**

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

def evaluate_model(name, y_true, y_pred):

    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average='weighted',
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average='weighted',
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average='weighted',
        zero_division=0
    )

    print("=" * 50)
    print(name)
    print("=" * 50)

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))

    return accuracy, precision, recall, f1
```

**Evaluate Logistic Regression**

```python
logistic_results = evaluate_model(
    "Logistic Regression",
    y_test,
    y_pred_logistic
)
```

**Evaluate Decision Tree**

```python
logistic_results = evaluate_model(
    "Logistic Regression",
    y_test,
    y_pred_logistic
)
```

**Evaluate Random Forest**

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

rf_results = evaluate_model(
    "Random Forest",
    y_test,
    y_pred_rf
)
```
